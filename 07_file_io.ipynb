```python
# Exo 6
import csv

#Read
with open("data/people.csv") as f:
    rows = list(csv.DictReader(f))

#Clean — try to convert age to int, drop bad rows
cleaned =[]
for row in rows:
    try:
        cleaned.append({
            "name": row["name"],
            "age": int(row["age"]),
            "city": row["city"],
        })
    except ValueError:
        print(f"Warning: skipping {row['name']} (bad age {row['age']!r})")

print(cleaned)

# Average age
ages = [r["age"] for r in cleaned]
print(f"Average age: {sum(ages) / len(ages):.1f}")

```

```python
# Exo 7
with open("data/cleaned_people.csv", "w", newline="") as f:
    writer =csv.DictWriter(f, fieldnames=["name", "age", "city"])
    writer.writeheader()
    writer.writerows(cleaned)
```

```python
with open("data/cleaned_people.csv", "r") as f:
    header = f.readline().strip().split(",")
    rows = []
    for line in f:
        values = line.strip().split(",")
        record = dict(zip(header, values))
        rows.append(record)

print(rows)
```

```python
with open("data/text.txt", "r") as f:
    content = f.read().lower().split()
    word_counts = {}
    for word in content:
        word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)

# Part two
sorted_counts = sorted(word_counts.items(), key=lambda kv: kv[1], reverse=True)

for word, count in sorted_counts[:10]:
    print(f"{word}: {count}")
```
