```python
def make_user(name, age, city="Paris"):
    return {"name": name, "age": age, "city": city}

make_user("Alice", 30)                                  # positional
make_user(name="Alice", age=30)                         # keyword
make_user(age=30, name="Alice")                         # keyword — order doesn't matter
make_user("Alice", city="Lyon", age=30)                 # mixed
```

```python
# Exo 3
def min_max_avg(numbers):
    """Return the minimum, maximum and the average of a  given list."""
    return min(numbers), max(numbers), sum(numbers) / len(numbers)
lista = [4, 8, 15, 16, 23, 42]
x, y, z = min_max_avg(lista)
print(f"min value: {x}, max value: {y}, average: {z}")
```

```python
# Exo 6
library = {
    "1984": {"author": "Orwell", "year": 1949, "available": True},
    "Dune": {"author": "Herbert", "year": 1965, "available": False},
    "Neuromancer": {"author": "Gibson", "year": 1984, "available": True},
}
def available_books(libr):
    books = []
    for title, info in libr.items():
        if info["available"]:
            books.append(title)

available_books(library)

def count_before_year(libr, year):
    count = 0
    for info in libr.values():
        if info["year"] < year:
            count += 1
    return count
# print(available_books(library))
print(count_before_year(library, 1960))
```

```python
# Exo 7
def word_count(text):
    """Returns a dictionary with the word key and the number of occurrences as a value."""
    word_counts = {}
    for word in text.split():
        word_counts[word] = word_counts.get(word, 0) + 1
    return word_counts

sentence = "the quick brown fox jumps over the lazy dog the fox"
print(word_count(sentence))
```

```python
# Exo 8 — Bonus
def clean_text(text):
    return text.strip().lower()
word_count(clean_text("  Hello World Hello"))
# clean_text lowercases and trims whitespace so that "Hello" and "hello"
# are counted as the same word. Without it, case would split the count.
```
