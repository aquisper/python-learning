```python
fruits2 = ["apple", "banana", "cherry", "date"]

fruits2.insert(1, "kiwi")      # insert at index 1 → ["apple", "kiwi", "banana", "cherry", "date"]
fruits2.remove("banana")       # remove first occurrence
popped = fruits2.pop()         # remove and return last element
fruits2[0] = "mango"           # change element at index 0

fruits2.sort()                 # sort in place (alphabetical/numerical)
fruits2.reverse()              # reverse in place
```

```python
#3 Sets— unordered, unique elements
# A set is a collection with no duplicates and no order. Written with curly braces.
unique_numbers = {1, 2, 3, 4}
colors = {"red", "green", "blue"}

# Duplicates are removed automatically
nums = {1, 1, 2, 2, 3, 3}
print(nums)
# nums = {1, 2, 3}

# Why sets?
# Fast membership checks. x in my_set is O(1), while x in my_list is O(n). For big data, sets are dramatically faster.
#
# Removing duplicates. set(my_list) is the fastest way to dedupe.
#
# Set operations: union, intersection, difference.
```

```python
# Common use in DS
# df["city"].unique()       # returns array of unique values
# len(set(column))          # count of distinct values
```

```python
#Operations with sets
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# a | b     # union: {1, 2, 3, 4, 5, 6}
# a & b     # intersection: {3, 4}
# a - b     # difference (in a. not in b): {1, 2}
a ^ b     # symmetric difference (in one but not both): {1, 2, 5, 6}
```

```python
#Exo 7
library = {
    "1984": {"author": "Orwell", "year": 1949, "available": True},
    "Dune": {"author": "Herbert", "year": 1965, "available": False},
    "Neuromancer": {"author": "Gibson", "year": 1984, "available": True},
}
# library["Dune"]["author"]
library["Dune"].get("author")
for book, value in library.items():
    if value["available"]:
        print(book)
#count books published before 1960
count = 0
for book, value in library.items():
    if value["year"] < 1960:
        count += 1
print(count)

```
