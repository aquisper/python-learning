```python
# Catch specific exceptions
try:
    x = int(input("Number: "))
    y = 10 / x
except ValueError:
    print("That's not a valid number.")
except ZeroDivisionError:
    print("Can't divide by zero.")
```

```python
# Multiple exception types in one except
try:
    value = int("abc")
except (ValueError, TypeError):
    print("Bad input")
```

```python
# Else and finally
try:
    result = 10 / int("2")
except ValueError:
    print("Bad input.")
except ZeroDivisionError:
    print("Can't divide by zero.")
else:
    print(f"Result: {result}")    # runs only if NO exception occurred
finally:
    print("Cleanup happens here.")  # ALWAYS runs, error or not
```

```python
# Raising own exceptions
def validate_age(age):
    if age < 0:
        raise ValueError(f"Age can't be negative: {age}")
    if age > 150:
        raise ValueError(f"Age is implausible: {age}")
    return age

print(validate_age(30))
validate_age(-5)
```

```python
# Exo 1
def safe_divide(a, b):
    try:
        return a/ b
    except ZeroDivisionError:
        return None
print(safe_divide(5, 3))
print(safe_divide(5, 0))
```

```python
# Exo 2
def safe_int(s):
    try:
        return int(s)
    except ValueError:
        return None
print(safe_int("42"))
print(safe_int("hello"))
print(safe_int("3.14"))
print(safe_int(""))
```

```python
# Exo 3
def safe_divide2(a, b):
    try:
        return a / b
    except (ZeroDivisionError, TypeError):
        return None
print(safe_divide2(10, 2))
print(safe_divide2(10, 0))
print(safe_divide2(10, "abc"))
```

```python
# Exo 4
def read_age(age_str):
    try:
        val = int(age_str)             # attempt the conversion
    except ValueError:
        return None                      # not a number → return None
    else:
        if 0 <= val <= 150:
            return val                 # valid age → return it
        raise ValueError(f"Age out of range: {age_str}")
    finally:
        print("read_age finished.")

print(read_age("25"))
print(read_age("abc"))
```

```python
# Exo 5
def validate_username(nom):
    """Validate a username: 3+ chars, no spaces."""
    if len(nom) < 3:
        raise ValueError(f"Name too short ({len(nom)} chars): {nom!r}")
    if " " in nom:
        raise ValueError(f"Name contains a space: {nom!r}")
    return True

for name in ["alice", "ab", "alice smith", "a b"]:
    try:
        # the !r in f-string means "use repr" — it wraps the value in quotes and adds also whitespace
        # ctrl +shift + u +2192 = →; ctrl +shift + u + 2014 = —
        print(f"{name!r} → {validate_username(name)}")
    except ValueError as e:
        print(f"{name!r} → {e}")
```

```python
# Exo 6
raw_values = ["42", "3.14", "hello", "", "100", "0.5", "N/A"]

def to_float(num):
    try:
        return float(num)
    except ValueError:
        return None

values = [v for v in (to_float(x) for x in raw_values) if v is not None]
print(values)
```

```python
# Exo 7
data = [
    {"name": "Alice", "age": "25"},
    {"name": "Bob", "age": "unknown"},
    {"name": "Carol", "age": "42"},
    {"name": "David", "age": ""},
]
cleaned = []
for record in data:
    try:
        age = int(record["age"])
        cleaned.append({"name": record["name"], "age": age})
    except ValueError:
        print(f"Warning: could not parse age for {record['name']}")


print(cleaned)

ages = [r["age"] for r in cleaned]
average_age = sum(ages) / len(ages)
print(f"Average age: {average_age:.1f}")

```
