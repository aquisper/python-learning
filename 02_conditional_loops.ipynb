```python
#Booleans and comparisons
print(5 > 3)
print(5 < 3)
```

```python
#if else
age = 17
if age >= 18:
    print("You're and adult.")
else:
    print("You're a minor.")
```

```python
#elif
score = 75
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >=60:
    print("D")
else:
    print("F")
```

```python
#and, or, not
age = 5
has_license = True

if age >= 18 and has_license:
    print("Can drive")

if age < 18 or not has_license:
    print("Cannot drive")
```

```python
#iteration over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}")
```

```python
#while loops
count = 0
while count < 5:
    print(count)
    count += 1
```

```python
#break and continue
for i in range(10):
    if i == 5:
        break
    print(i)
print()
for i in range(5):
    if i == 2:
        continue
    print(i)
```

```python
#Combining loops and conditionals
numbers = [3, 12, 7, 20, 5, 8]
for n in numbers:
    if n > 10:
        print(f"{n} is big")
    else:
        print(f"{n} is small")
```

```python
#Exo 1: Even or odd
this_number = 17
if this_number % 2 == 0:
    print(f"{this_number} is even")
else:
    print(f"{this_number} is odd")
```

```python
#Exo 2 Grade classifier
score = 82
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >=60:
    print("D")
else:
    print("F")
```

```python
#Exo 3 FizzBuzz
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
#Exo 4 Sum and average
numbers = [4, 8, 15, 16, 23, 42]
suma = 0

for i in numbers:
    suma += i
print(suma)
average = suma / len(numbers)
print(average)
```

```python
#Exo 5 Countdown
counter = 10
while counter >= 1:
    print(counter)
    counter -= 1
print("Liftoff!")
```

```python
#Exo 6 Primes under 20
for n in range(2, 21):
    is_prime = True
    for i in range(2, n):
        if n % i == 0:
            # n has a divisor → not prime
            is_prime = False
            break
    if is_prime:
        print(n)
```
