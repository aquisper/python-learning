```python
# 1 The problem comprehension solve
squares = [n ** 2 for n in range(10)]
print(squares)
```

```python
# 2 The list comprehension general form
# [ expression for item in iterable if condition]
```

```python
evens = [ n for n in range(20) if n % 2 == 0]
print(evens)
```

```python
even_squares = [ n ** 2 for n in range(20) if n % 2 == 0]
print(even_squares)
```

```python
# 4 — on strings
words = ["hello", "world", "python", "data"]
lengths = [ len(w) for w in words]
print(lengths)
```

```python
uppercase = [w.upper() for w in words]
print(uppercase)
```

```python
long_words = [w for w in words if len(w) > 4]
print(long_words)
```

```python
# the equivalent loop to 4
lengths_for = []
for w in words:
    lengths_for.append(len(w))
print(lengths_for)
```

```python
# 6 — with and if/else in the expression
result = ["even" if n % 2 == 0 else "odd" for n in range(6)]
print(result)
```

```python
# Nested loops
pairs = [(x, y) for x in range(3) for y in range(3)]
print(pairs)
```

```python
# Same things, but with nested loops
pairs_loop = []
for x in range(3):
    for y in range(3):
        pairs_loop.append((x, y))
print(pairs_loop)
```

```python
matrix = [[1, 2, 3], [4, 5 ,6], [7, 8, 9]]
flat = [n for row in matrix for n in row]
print(flat)
```

```python
# Dict comprehension
squares_dict = {n: n ** 2 for n in range(6)}
print(squares_dict)
```

```python
# Equivalent loop
squares_dict_loop = {}
for n in range(6):
    squares_dict_loop[n] = n ** 2
print(squares_dict_loop)
```

```python
# Example — word lengths
words = ["hello","world", "python"]
lengths = {w: len(w) for w in words}
print(lengths)
```

```python
# Example — filter a dict
prices = {"apple": 1.5, "banana": 0.8, "cherry": 3.0, "date": 5.5}
expensive = {k: v for k, v in prices.items() if v > 2}
print(expensive)
```

```python
# 5 Set comprehension
unique_lengths = {len(w) for w in ["hello", "world", "python", "data"]}
print(unique_lengths)

words = ["hello", "world", "Hello", "data", "WORLD"]
unique_lower = {w.lower() for w in words}
```

```python
# Exo 10
library = {
    "1984": {"author": "Orwell", "year": 1949, "available": True},
    "Dune": {"author": "Herbert", "year": 1965, "available": False},
    "Neuromancer": {"author": "Gibson", "year": 1984, "available": True},
}
def available_books(libr):
    return [title for title, info in libr.items() if info["available"]]
available_books(library)
# for consistency and less code, I prefer this solution.
```
